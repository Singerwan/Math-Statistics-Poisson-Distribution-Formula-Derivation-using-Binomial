<hr>
premises: convert the probability into either occur or not occur and on top of that <br>
in order to elminate the probability of multiple events occur simultaneously or within a certain time interval,<br>
we assume n will approache infinity which is to subdivid interval into subintervals, reducing the chances of multiple events occuring within a fix interval. <br>
purpose : 
<br>
&emsp;&emsp;to let <mark>n</mark>
(as its the approximation using binomial then n is not the number of trials instead its the number of <mark>subintervals</mark> )
<br>&emsp;&emsp;to approaches infinity is to allow error to approach zero
